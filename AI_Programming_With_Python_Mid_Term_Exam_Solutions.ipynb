```python
# You must run this code and use this dataset for all questions

import numpy as np
import pandas as pd

np.random.seed(24)

num_employees = 50  # you can increase this if needed

# Employee data
emp_ids = np.arange(101, 101 + num_employees)
emp_ages = np.random.randint(20, 55, num_employees)
emp_income = np.random.randint(35000, 95000, num_employees)

# Fixed set of roles (repetition allowed)
roles_list = [
    "HR Manager",
    "Software Developer",
    "Accountant",
    "Marketing Executive",
    "Sales Executive",
    "Backend Developer",
    "HR Executive",
    "Financial Analyst",
    "Digital Marketer",
    "Operations Manager"
]

emp_roles = np.random.choice(roles_list, num_employees)

# Create DataFrame
data = pd.DataFrame({
    "emp_id": emp_ids,
    "emp_age": emp_ages,
    "emp_salary": emp_income,
    "role": emp_roles
})

# Save dataset
data.to_csv("staff_data.csv", index=False)
```

Q4.
Create a new NumPy array where each employee’s age is adjusted based on:
* Add 4 years if age is less than the average age
* Add 2 years if age is greater than or equal to the average age

Do NOT modify the original array.

Print:
* original array
* updated array
* average age used for condition


```python
avg_age = np.average(emp_ages)
new_emp_ages = []
for x in emp_ages:
  if(x<avg_age):
    x+=4
    new_emp_ages.append(x)
  elif(x>=avg_age):
    x+=2
    new_emp_ages.append(x)
new_emp_ages = np.array(new_emp_ages)

print(emp_ages) #original array
print(new_emp_ages) #updated array
print(avg_age) #average age used for condition
```

Q5.
Using NumPy:
* Assume all salaries of Tech roles increase by 8%
Calculate the new total salary expense
* Find the overall increase in expense after the raise

 Print both old and new totals along with the difference.


```python
new_emp_salary = []

for x in range(emp_income.size):
  if emp_roles[x] == 'Software Developer':
    new_emp_salary.append(emp_income[x] + 0.08*emp_income[x])

  elif emp_roles[x]=='Backend Developer':
    new_emp_salary.append(emp_income[x] + 0.08*emp_income[x])

  elif emp_roles[x]== 'Digital Marketer':
    new_emp_salary.append(emp_income[x] + 0.08*emp_income[x])

  else:
    new_emp_salary.append(emp_income[x])

new_emp_salary = np.array(new_emp_salary)
old_expense = np.sum(emp_income)
new_expense = np.sum(new_emp_salary)
difference = new_expense - old_expense



print(new_emp_salary)

print(old_expense)
print(new_expense)
print(difference)
```

Q10.Using .loc:
* Increase salary by 12% for "Software Developer" and "Backend Developer"
 * Increase salary by 5% for all other roles

Then:
* Print the dataframe that includes role , previous_salary , new_salary , increase_of_salary


```python
new_salary_df = df.copy()

conditions = ((new_salary_df['role']=='Software Developer') |(new_salary_df['role']=='Backend Developer') )
new_salary_df['emp_salary'] = new_salary_df['emp_salary'].astype(float)
new_salary_df.loc[conditions,'emp_salary']= new_salary_df['emp_salary']* 1.12
new_salary_df.loc[~conditions,'emp_salary']= new_salary_df['emp_salary']* 1.05

new_salary_df['previous_salary']=df['emp_salary']

new_salary_df['increase_of_salary']=new_salary_df['emp_salary']-new_salary_df['previous_salary']

new_salary_df.rename(columns={'emp_salary':'new_salary'},inplace=True)

new_df = new_salary_df.loc[ : ,  ['role','previous_salary','new_salary','increase_of_salary']]

print(new_df)











```
