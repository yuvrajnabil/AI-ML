# **Lesson 2: __init__() Constructor**

```python
# Task 1
# Student নামে class বানাও যেখানে:

# constructor এ name আর roll নেবে
# show_data() method থাকবে
# 2টা object বানাবে

class Student:
  def __init__(self,name,roll):
    self.name= name
    self.roll= roll
  def show_data(self):
    print(f"Name is {self.name}")
    print(f"Roll is {self.roll}")

s1=Student("Nabil",73)
s2=Student("Jui",40)
s1.show_data()
s2.show_data()

# Task 2
# 3–4 line-এ explain করো:

# set_data() আর __init__() এর difference কী
#  1. set_data() call korte hoi manually  r __init__ e automaticaly call hoi input hoi object creat er smy.
#  2.input hoi object creat er por dite hoi r __init__ e obeject create er smy dite hoi.
#  3.set_data() use code clean hoi na r real project use kom __init__ e code clean thake r real project use beshi

# Task 3
# এই code-এর output কী হবে বলো:
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author

    def show_data(self):
        print(self.title)
        print(self.author)

b1 = Book("OOP Basics", "Anik")
b1.show_data()

# Output hobe:
# OOP Basics
# Anik
```

# **Lesson 3: Default Constructor vs Parameterized Constructor**

```python
# Task 1
# Book নামে ekta class banao jekhane default constructor thakbe।

# Requirements:

# title = "Unknown"
# author = "Unknown"
# show_data() method thakbe
# 1 ta object create kore data show korbe
class Book:
  def __init__(self):
    self.title = "Unknown"
    self.author = "Unknown"
  def show_data(self):
    print(f"Title is {self.title}")
    print(f"Author is {self.author}")
book1 = Book()
book1.show_data()

# Task 2
# Car নামে ekta class banao jekhane parameterized constructor thakbe।

# Requirements:

# brand
# color
# show_data() method
# 2 ta object create korbe with different data

class Car:
  def __init__(self,brand,color):
    self.brand = brand
    self.color = color
  def show_data(self):
    print(f"Brand name is {self.brand}")
    print(f"Color is {self.color}")
car1 = Car("Audi","Black")
car2 = Car("BMW","Blue")
car1.show_data()
car2.show_data()

# Task 3
# 3–4 line-e explain koro:

# default constructor
# parameterized constructor
# kon situation-e konta use kora better
# Default constructor bolte bujhai emon constructor jekhane object create korar shomoy baire theke value na dileo cholbe
# Parameterized constructor holo sheta jekhane object create korar shomoy value pass korte hoy।
# Jodi problem-e bole onek book, onek student, onek car with different info, tahole default fixed value use korle thik hobe na.
# Tokhon parameterized constructor better.
# create object with default values initially set unknown, 0, empty, none no input during creation Tahole default constructor use kora jete pare.

# Task 4
# Nicher code-er output bolo:
class Student:
    def __init__(self, name="Unknown", roll=0):
        self.name = name
        self.roll = roll

    def show_data(self):
        print(self.name)
        print(self.roll)

s1 = Student()
s2 = Student("Jui", 40)

s1.show_data()
s2.show_data()
# output:
# Unknown
# 0
# Jui
# 40
```

# **Lesson 5: Instance Method and Data Update**

Aaj amra shikhbo:

* Instance method
* method diye object-er data update kora
* same object-e multiple method use kora




Eta bujhle tumi class diye shudhu data store na, data change-o korte parba ✨



```python
# Instance Method ki?
# Je method object-er sathe kaj kore, ar self use kore object-er data access ba change kore, take instance method bole।

# Task 1
# Student নামে ekta class banao।

# Requirements:

# constructor e name and roll nibe
# show_data() method thakbe
# update_roll(new_roll) method thakbe
# 1 ta object create kore age data show korbe
# then roll update kore abar data show korbe

class Student:
  def __init__(self,name,roll):
    self.name = name
    self.roll = roll
  def update_roll(self,new_roll):
    self.roll = new_roll
  def show_data(self):
    print(f"Student name is {self.name}")
    print(f"{self.name}'s roll is {self.roll}")
s1=Student("Nabil",73)
s1.show_data()
s1.update_roll(80)
s1.show_data()

# Task 2
# Car নামে ekta class banao।

# Requirements:

# constructor e brand and color
# show_data() method
# update_color(new_color) method
# 1 ta object create kore color update korbe

class Car:
  def __init__(self,brand,color):
    self.brand = brand
    self.color = color
  def show_data(self):
    print(f"Car brand is {self.brand}")
    print(f"Color of the car is {self.color}")
  def update_color(self,new_color):
    self.color = new_color
car1= Car("Audi", "Red")
car1.show_data()
car1.update_color("Black")
car1.show_data()

# Task 3
# 3–4 line-e explain koro:

# instance method ki
# data update korte method keno use kora hoy
# answer:Instance method holo oi method je object-er data niye kaj kore ebong self use kore।
# Eta diye object-er data dekhano ba update kora jay।
# Real project-e object create-r por onek shomoy data change korte hoy।
# Tai update korar jonno method use kora hoy।


#  Task 4
# Nicher code-er output bolo:
class Book:
    def __init__(self, title):
        self.title = title

    def update_title(self, new_title):
        self.title = new_title

    def show_data(self):
        print(self.title)

b1 = Book("Python")
b1.update_title("OOP in Python")
b1.show_data()

# Answe: OOP in Python
```

# **Lesson 6: Accessing Object Data Outside the Class**

Aaj amra shikhbo:

* object-er data class-er baire theke access kora
* dot notation
* object-er attribute read kora
* object-er attribute change kora
* direct access vs method use

Ei concept ta khub important, karon object bananor por amra onek shomoy baire theke tar data dekhte ba change korte chai ✨

```python
# Task 1
# Student নামে ekta class banao।

# Requirements:

# constructor e name and roll
# 1 ta object create korbe
# class-er baire theke name and roll print korbe
class Student:
  def __init__(self,name,roll):
    self.name = name
    self.roll = roll
s1= Student("Nabil",73)
print(f"Student name is {s1.name}")
print(f"{s1.name}'s roll is {s1.roll}")

# Task 2
# Car নামে ekta class banao।

# Requirements:

# constructor e brand and color
# object create korbe
# direct access diye color change korbe
# change-r age and pore print korbe

class Car:
  def __init__(self,brand,color):
    self.brand = brand
    self.color = color
car1= Car("BMW","Red")
print(f"Car brand is {car1.brand}")
print(f"Color of the car is {car1.color}")
car1.color="Black"
print(f"Car brand is {car1.brand}")
print(f"Color of the car is {car1.color}")


# Task 3
# 3–4 line-e explain koro:

# dot notation ki
# direct access diye ki kora jay
# method use kora keno better hote pare

# Answer: dot notation holo Object-er kono variable ba method access korte hole amra dot (.) use kori
# direct access diye ki kora jay direct data show kora jai, Data change kora jai method use na korei,
# Method use kora better karon
# better control dey
# update-r age check kora jay
# real project-e beshi useful

# Task 4
# Nicher code-er output bolo:

class Book:
    def __init__(self, title):
        self.title = title

b1 = Book("Python")
print(b1.title)
b1.title = "OOP"
print(b1.title)
```
